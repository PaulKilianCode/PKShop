```python
%%sql
SELECT subcategorykey,
COUNT(DISTINCT CASE WHEN unitprice < 1000 THEN orderkey END) AS orders_below1000,
COUNT(DISTINCT CASE WHEN unitprice BETWEEN 1000 AND 2000 THEN orderkey END) AS orders_between1000and2000,
COUNT(DISTINCT CASE WHEN unitprice > 1000 THEN orderkey END) AS orders_greater2000
FROM sales
LEFT JOIN product on sales.productkey = product.productkey
WHERE EXTRACT(YEAR from orderdate) = 2023
GROUP BY subcategorykey
ORDER BY subcategorykey
LIMIT 10
```
